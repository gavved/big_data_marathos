```python
%sql
-- View for athlete performances in time-type events
create or replace view marathos.gold.view_time_events_athletes as

select
    f.result_id,
    f.performance_seconds,
    f.athlete_average_speed,

    a.athlete_id,
    a.athlete_country,
    a.athlete_gender,
    a.athlete_age_category,

    e.event_name,
    e.event_dates,
    e.event_distance_length,
    e.event_type,

    c.athlete_club

from marathos.gold.fct_results f
join marathos.gold.dim_event e on f.event_id = e.event_id
join marathos.gold.dim_athlete a on f.athlete_id = a.athlete_id
join marathos.gold.dim_club c on f.club_id = c.club_id
where e.event_type = 'Time'
```

```python
%sql
-- View for time-type events statistics
create or replace view marathos.gold.view_distance_events as

select
    e.event_id,
    e.event_name,
    e.event_dates,
    e.event_distance_length,
    e.event_number_of_finishers,
    e.event_type,

    count(f.result_id) as participants

from marathos.gold.fct_results f
join marathos.gold.dim_event e on f.event_id = e.event_id
where e.event_type = 'Time'
group by 
    e.event_id,
    e.event_name,
    e.event_dates,
    e.event_distance_length,
    e.event_number_of_finishers,
    e.event_type
```

```python
## Read data
raw_data = spark.read.csv("/Volumes/marathos/raw/raw_data/TWO_CENTURIES_OF_UM_RACES.csv",
                          header=True,
                          inferSchema=True)
```

```python
## Overview
display(raw_data)
```

```python
## Number of rows
raw_data.count()
```

```python
## Number of columns
len(raw_data.columns)
```

```python
## Check the inferred schema
raw_data.printSchema()
```

```python
## Descriptive summary of numerical fields
raw_data.describe().show()
```

```python
## Count nulls

from pyspark.sql.functions import col, sum

raw_data.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in raw_data.columns
]).show()
```

```python
## Unique events
raw_data.select('Event name').distinct().count()
```

```python
## Unique events where the same event on different dates is considered as a different event
raw_data.select("Event name", "Event dates").distinct().count()
```

```python
## Unique athletes
raw_data.select('Athlete ID').distinct().count()
```

```python
## Unique countries
raw_data.select('Athlete country').distinct().count()
```

```python
## Age distribution
raw_data.groupBy('Athlete age category').count().orderBy('count', ascending=False).show()
```

```python
## Country distribution
raw_data.groupBy('Athlete country').count().orderBy('count', ascending=False).show()
```
